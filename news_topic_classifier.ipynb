```python
from transformers import pipeline
classifier = pipeline(
    "zero-shot-classification",
    model = "MoritzLaurer/mDeBERTa-v3-base-mnli-xnli"
)
```

```python
labels = ['sports', 'politics', 'technology', 'business', 'health', 'finance', 'education', 'entertainment']
```

```python
article = """
Apple announced new AI-powered chips for future Mac devices
during its annual technology event in California.
"""

result = classifier(article, candidate_labels=labels)
top_label = result['labels'][0]
top_score = result['scores'][0]

print(f"Predicted Category: {top_label}")
print(f"Confidence: {top_score*100:.2f}%")
```

```python
while True:
  article = input('\nEnter news article text (or type exit) ')

  if article.lower() == 'exit':
    print("Exiting News Topic Classifier...")
    break

  result = classifier(article, candidate_labels=labels)
  top_label = result['labels'][0]
  top_score = result['scores'][0]

  print(f"Predicted Category: {top_label}")
  print(f"Confidence: {top_score*100:.2f}%")
```

```python
sample_articles = [
    "The football team won the championship after a thrilling final match.",

    "The government announced a new economic reform policy in parliament.",

    "Researchers developed a new AI model for medical image analysis.",

    "Stock markets saw major growth after strong quarterly earnings reports."
]

for article in sample_articles:

    result = classifier(article, candidate_labels=labels)

    top_label = result["labels"][0]
    top_score = result["scores"][0]

    print("\nArticle:")
    print(article)

    print(f"\nPredicted Category: {top_label}")
    print(f"Confidence: {top_score * 100:.2f}%")
```
